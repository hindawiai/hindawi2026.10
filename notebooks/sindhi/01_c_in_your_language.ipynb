```python
%%hindawi_c sindhi
#ضمن <stdio.h>
/* quick sort */
فراغ مٽا_سٽا(صحيح *پهريون, صحيح *ٻيو) { صحيح عارضي = *پهريون; *پهريون = *ٻيو; *ٻيو = عارضي; }
صحيح حصا_ڪريو(صحيح قدرون[], صحيح هيٺيون, صحيح مٿاهون) {
  صحيح محور = قدرون[مٿاهون], اشاريو = هيٺيون - 1;
  لكل (صحيح جاچ = هيٺيون; جاچ < مٿاهون; جاچ++)
    إذا (قدرون[جاچ] <= محور) { اشاريو++; مٽا_سٽا(&قدرون[اشاريو], &قدرون[جاچ]); }
  مٽا_سٽا(&قدرون[اشاريو + 1], &قدرون[مٿاهون]);
  أرجع اشاريو + 1;
}
فراغ تيز_ترتيب_ڏيو(صحيح قدرون[], صحيح هيٺيون, صحيح مٿاهون) {
  إذا (هيٺيون < مٿاهون) { صحيح وچ = حصا_ڪريو(قدرون, هيٺيون, مٿاهون); تيز_ترتيب_ڏيو(قدرون, هيٺيون, وچ - 1); تيز_ترتيب_ڏيو(قدرون, وچ + 1, مٿاهون); }
}
صحيح مکيه(فراغ) {
  صحيح قدرون[] = {29, 3, 72, 14, 8, 55, 1, 40};
  تيز_ترتيب_ڏيو(قدرون, 0, 7);
  لكل (صحيح اشاريو = 0; اشاريو < 8; اشاريو++) لکو("%d ", قدرون[اشاريو]);
  لکو("\n");
  أرجع 0;
}
```
