```python
#@title **Your Yacc program, in ਪੰਜਾਬੀ**
#@markdown Double-click to edit, then run the next cell.
%%writefile /content/hin/calculator.y
%{
#ਅਨ੍ਤਰ੍ਭਾਵ <stdio.h>
#ਅਨ੍ਤਰ੍ਭਾਵ <ctype.h>
ਪੂਰਨ yylex(ਖਾਲੀ);
ਖਾਲੀ yyerror(ਅਚਲ ਅੱਖਰ *s) { fprintf(stderr, "%s\n", s); }
%}
%token NUM
%ਖੱਬੇ '+'
%ਖੱਬੇ '*'
%%
input: | input ਸਤਰ ;
ਸਤਰ: '\n' | expr '\n' { ਲਿਖੋ("%d\n", $1); } ;
expr: NUM | expr '+' expr { $$ = $1 + $3; } | expr '*' expr { $$ = $1 * $3; } ;
%%
ਪੂਰਨ yylex(ਖਾਲੀ) {
  ਪੂਰਨ c = getchar();
  ਜਦੋਂ (c == ' ') c = getchar();
  ਜੇ (ਅੰਕ_ਹੈ(c)) { yylval = c - '0'; ਜਦੋਂ (ਅੰਕ_ਹੈ(c = getchar())) yylval = yylval * 10 + c - '0'; ungetc(c, stdin); ਵਾਪਸ NUM; }
  ਵਾਪਸ c == EOF ? 0 : c;
}
ਪੂਰਨ ਮੁੱਖ(ਖਾਲੀ) { ਵਾਪਸ yyparse(); }
```

# Example in English (for reference)
The same programs in English, to compare and to cross-check.

```python
#@title **C in English**
%%writefile /content/hin/english_binary_search.c
#include <stdio.h>
/* binary search in a sorted array */
int binary_search(int values[], int count, int key) {
  int low = 0, high = count - 1;
  while (low <= high) {
    int middle = low + (high - low) / 2;
    if (values[middle] == key) return middle;
    if (values[middle] < key) low = middle + 1; else high = middle - 1;
  }
  return -1;
}
int main(void) {
  int values[] = {2, 5, 8, 12, 16, 23, 38, 56, 72, 91};
  printf("%d %d\n", binary_search(values, 10, 23), binary_search(values, 10, 7));
  return 0;
}
```

```python
#@title **C++ in English**
%%writefile /content/hin/english_bst.cpp
// a binary search tree class, in C++
#include <iostream>
class Tree {
  struct Node { int key; Node *left = nullptr, *right = nullptr; };
  Node *root = nullptr;
  Node *insert(Node *node, int key) { if (!node) { node = new Node; node->key = key; return node; } if (key < node->key) node->left = insert(node->left, key); else node->right = insert(node->right, key); return node; }
  void inorder(Node *node) { if (!node) return; inorder(node->left); std::cout << node->key << " "; inorder(node->right); }
public:
  void insert(int key) { root = insert(root, key); }
  void print() { inorder(root); std::cout << std::endl; }
};
int main() { Tree tree; int keys[] = {5, 3, 8, 1, 4}; for (int key : keys) tree.insert(key); tree.print(); return 0; }
```

```python
#@title **Python in English**
%%writefile /content/hin/english_binary_search.py
# binary search, in Python
def binary_search(values, key):
    low, high = 0, len(values) - 1
    while low <= high:
        middle = (low + high) // 2
        if values[middle] == key:
            return middle
        if values[middle] < key:
            low = middle + 1
        else:
            high = middle - 1
    return -1

print(binary_search([2, 5, 8, 12, 16, 23, 38], 23), binary_search([2, 5, 8], 7))
```

```python
#@title **Yacc in English**
%%writefile /content/hin/english_calculator.y
%{
#include <stdio.h>
#include <ctype.h>
int yylex(void);
void yyerror(const char *s) { fprintf(stderr, "%s\n", s); }
%}
%token NUM
%left '+'
%left '*'
%%
input: | input line ;
line: '\n' | expr '\n' { printf("%d\n", $1); } ;
expr: NUM | expr '+' expr { $$ = $1 + $3; } | expr '*' expr { $$ = $1 * $3; } ;
%%
int yylex(void) {
  int c = getchar();
  while (c == ' ') c = getchar();
  if (isdigit(c)) { yylval = c - '0'; while (isdigit(c = getchar())) yylval = yylval * 10 + c - '0'; ungetc(c, stdin); return NUM; }
  return c == EOF ? 0 : c;
}
int main(void) { return yyparse(); }
```
