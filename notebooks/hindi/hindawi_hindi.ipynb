## Detailed description

### Shaili guru: C in Hindi

Shaili guru is C written in the language: the language of systems, memory and the machine, for high school, undergraduate and GATE students. A program here goes through the script filter and Romenagri, the language's transducer gives C, and gcc compiles it; the names reach the binary in their Romenagri form.

**Program editing.** Edit the program in the next cell, then run it to save it.

```python
%%writefile program_guru.c
#समावेश <stdio.h>
/* binary search in a sorted array */
पूर्णांक binary_search(पूर्णांक values[], पूर्णांक count, पूर्णांक कुंजी) {
  पूर्णांक low = 0, high = count - 1;
  जबतक (low <= high) {
    पूर्णांक middle = low + (high - low) / 2;
    अगर (values[middle] == कुंजी) वापस middle;
    अगर (values[middle] < कुंजी) low = middle + 1; अन्यथा high = middle - 1;
  }
  वापस -1;
}
पूर्णांक मुख्य(व्योम) {
  पूर्णांक values[] = {2, 5, 8, 12, 16, 23, 38, 56, 72, 91};
  म_लिखो("%d %d\n", binary_search(values, 10, 23), binary_search(values, 10, 7));
  वापस 0;
}
```

**Input editing.** Anything the program reads goes here.

```python
#@title Input editing
INPUT = """"""   # what the program reads, one value per line
open('input_guru.txt', 'w', encoding='utf-8').write(INPUT)
```

**The compilation flow.** The program as written, the script filter's Devanagari (for scripts that need one), the Romenagri form the compiler works on, and the host-language text.

```python
nb.show_stages(open('program_guru.c', encoding='utf-8').read(), 'hindi', 'guru')
```

**Program execution.**

```python
nb.build_and_run(open('program_guru.c', encoding='utf-8').read(), 'hindi', 'guru',
                 open('input_guru.txt', encoding='utf-8').read())
```

**From the host language (Std2Hin).** The English original of the same program, brought into Hindi by BURP.

```python
print(nb.from_host(open(REPO + '/samples/gate/algorithms/searching/binary_search.c', encoding='utf-8').read(), 'hindi', 'guru'))
```

### Shaili soochee: Python in Hindi

Shaili soochee is Python written in the language. It goes through the same path as C, with the language's Python transducer at the end, so a name has the same form in Python as in C.

**Program editing.** Edit the program in the next cell, then run it to save it.

```python
%%writefile program_soochee.py
# binary search, in Python
def binary_search(values, कुंजी):
    low, high = 0, len(values) - 1
    जबतक low <= high:
        middle = (low + high) // 2
        अगर values[middle] == कुंजी:
            वापस middle
        अगर values[middle] < कुंजी:
            low = middle + 1
        अन्यथा:
            high = middle - 1
    वापस -1

print(binary_search([2, 5, 8, 12, 16, 23, 38], 23), binary_search([2, 5, 8], 7))
```

**Input editing.** Anything the program reads goes here.

```python
#@title Input editing
INPUT = """"""   # what the program reads, one value per line
open('input_soochee.txt', 'w', encoding='utf-8').write(INPUT)
```

**The compilation flow.** The program as written, the script filter's Devanagari (for scripts that need one), the Romenagri form the compiler works on, and the host-language text.

```python
nb.show_stages(open('program_soochee.py', encoding='utf-8').read(), 'hindi', 'soochee')
```

**Program execution.**

```python
nb.build_and_run(open('program_soochee.py', encoding='utf-8').read(), 'hindi', 'soochee',
                 open('input_soochee.txt', encoding='utf-8').read())
```

**From the host language (Std2Hin).** The English original of the same program, brought into Hindi by BURP.

```python
print(nb.from_host(open(REPO + '/samples/gate/paradigms/binary_search.py', encoding='utf-8').read(), 'hindi', 'soochee'))
```
