## Detailed description

### Shaili guru: C in Telugu

Shaili guru is C written in the language: the language of systems, memory and the machine, for high school, undergraduate and GATE students. A program here goes through the script filter and Romenagri, the language's transducer gives C, and gcc compiles it; the names reach the binary in their Romenagri form.

**Program editing.** Edit the program in the next cell, then run it to save it.

```python
%%writefile program_guru.c
#అన్తర్భావ <stdio.h>
/* binary search in a sorted array */
పూర్ణాంకం ద్విమాన_శోధన(పూర్ణాంకం విలువలు[], పూర్ణాంకం లెక్క, పూర్ణాంకం కీలకం) {
  పూర్ణాంకం దిగువ = 0, ఎగువ = లెక్క - 1;
  అయితే (దిగువ <= ఎగువ) {
    పూర్ణాంకం మధ్య = దిగువ + (ఎగువ - దిగువ) / 2;
    ఒకవేళ (విలువలు[మధ్య] == కీలకం) తిరిగి మధ్య;
    ఒకవేళ (విలువలు[మధ్య] < కీలకం) దిగువ = మధ్య + 1; లేకపోతే ఎగువ = మధ్య - 1;
  }
  తిరిగి -1;
}
పూర్ణాంకం ప్రధాన(శూన్య) {
  పూర్ణాంకం విలువలు[] = {2, 5, 8, 12, 16, 23, 38, 56, 72, 91};
  రాయి("%d %d\n", ద్విమాన_శోధన(విలువలు, 10, 23), ద్విమాన_శోధన(విలువలు, 10, 7));
  తిరిగి 0;
}
```

**Input editing.** Anything the program reads goes here.

```python
#@title Input editing
INPUT = """"""   # what the program reads, one value per line
open('input_guru.txt', 'w', encoding='utf-8').write(INPUT)
```

**The compilation flow.** The program as written, the script filter's Devanagari (for scripts that need one), the Romenagri form the compiler works on, and the host-language text.

```python
nb.show_stages(open('program_guru.c', encoding='utf-8').read(), 'telugu', 'guru')
```

**Program execution.**

```python
nb.build_and_run(open('program_guru.c', encoding='utf-8').read(), 'telugu', 'guru',
                 open('input_guru.txt', encoding='utf-8').read())
```

**From the host language (Std2Hin).** The English original of the same program, brought into Telugu by BURP.

```python
print(nb.from_host(open(REPO + '/samples/gate/algorithms/searching/binary_search.c', encoding='utf-8').read(), 'telugu', 'guru'))
```

### Shaili soochee: Python in Telugu

Shaili soochee is Python written in the language. It goes through the same path as C, with the language's Python transducer at the end, so a name has the same form in Python as in C.

**Program editing.** Edit the program in the next cell, then run it to save it.

```python
%%writefile program_soochee.py
# binary search, in Python
నిర్వచనం ద్విమాన_శోధన(విలువలు, కీలకం):
    దిగువ, ఎగువ = 0, పొడవు(విలువలు) - 1
    అయితే దిగువ <= ఎగువ:
        మధ్య = (దిగువ + ఎగువ) // 2
        ఒకవేళ విలువలు[మధ్య] == కీలకం:
            తిరిగి మధ్య
        ఒకవేళ విలువలు[మధ్య] < కీలకం:
            దిగువ = మధ్య + 1
        లేకపోతే:
            ఎగువ = మధ్య - 1
    తిరిగి -1

ముద్రించు(ద్విమాన_శోధన([2, 5, 8, 12, 16, 23, 38], 23), ద్విమాన_శోధన([2, 5, 8], 7))
```

**Input editing.** Anything the program reads goes here.

```python
#@title Input editing
INPUT = """"""   # what the program reads, one value per line
open('input_soochee.txt', 'w', encoding='utf-8').write(INPUT)
```

**The compilation flow.** The program as written, the script filter's Devanagari (for scripts that need one), the Romenagri form the compiler works on, and the host-language text.

```python
nb.show_stages(open('program_soochee.py', encoding='utf-8').read(), 'telugu', 'soochee')
```

**Program execution.**

```python
nb.build_and_run(open('program_soochee.py', encoding='utf-8').read(), 'telugu', 'soochee',
                 open('input_soochee.txt', encoding='utf-8').read())
```

**From the host language (Std2Hin).** The English original of the same program, brought into Telugu by BURP.

```python
print(nb.from_host(open(REPO + '/samples/gate/paradigms/binary_search.py', encoding='utf-8').read(), 'telugu', 'soochee'))
```
