```python
%%hindawi_c urdu
#شامل <stdio.h>
/* quick sort */
خالی تبادلہ(صحیح *پہلا, صحیح *دوسرا) { صحیح عارضی = *پہلا; *پہلا = *دوسرا; *دوسرا = عارضی; }
صحیح تقسیم(صحیح اقدار[], صحیح کم, صحیح بلند) {
  صحیح محور = اقدار[بلند], اشاریہ = کم - 1;
  برائے (صحیح جائزہ = کم; جائزہ < بلند; جائزہ++)
    اگر (اقدار[جائزہ] <= محور) { اشاریہ++; تبادلہ(&اقدار[اشاریہ], &اقدار[جائزہ]); }
  تبادلہ(&اقدار[اشاریہ + 1], &اقدار[بلند]);
  واپس اشاریہ + 1;
}
خالی تیز_ترتیب_دو(صحیح اقدار[], صحیح کم, صحیح بلند) {
  اگر (کم < بلند) { صحیح درمیان = تقسیم(اقدار, کم, بلند); تیز_ترتیب_دو(اقدار, کم, درمیان - 1); تیز_ترتیب_دو(اقدار, درمیان + 1, بلند); }
}
صحیح مین(خالی) {
  صحیح اقدار[] = {29, 3, 72, 14, 8, 55, 1, 40};
  تیز_ترتیب_دو(اقدار, 0, 7);
  برائے (صحیح اشاریہ = 0; اشاریہ < 8; اشاریہ++) لکھو("%d ", اقدار[اشاریہ]);
  لکھو("\n");
  واپس 0;
}
```
