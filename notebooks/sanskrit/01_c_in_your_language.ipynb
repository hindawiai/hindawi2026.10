```python
%%hindawi_c sanskrit
#अन्तर्भाव <stdio.h>
/* quick sort */
शून्य विनिमय(पूर्णाङ्क *प्रथमम्, पूर्णाङ्क *द्वितीयम्) { पूर्णाङ्क अस्थायी = *प्रथमम्; *प्रथमम् = *द्वितीयम्; *द्वितीयम् = अस्थायी; }
पूर्णाङ्क विभाजय(पूर्णाङ्क मूल्यानि[], पूर्णाङ्क निम्नम्, पूर्णाङ्क उच्चम्) {
  पूर्णाङ्क धुरी = मूल्यानि[उच्चम्], अनुक्रमः = निम्नम् - 1;
  प्रति (पूर्णाङ्क अवलोकनम् = निम्नम्; अवलोकनम् < उच्चम्; अवलोकनम्++)
    यदि (मूल्यानि[अवलोकनम्] <= धुरी) { अनुक्रमः++; विनिमय(&मूल्यानि[अनुक्रमः], &मूल्यानि[अवलोकनम्]); }
  विनिमय(&मूल्यानि[अनुक्रमः + 1], &मूल्यानि[उच्चम्]);
  प्रत्यागम अनुक्रमः + 1;
}
शून्य शीघ्र_क्रमय(पूर्णाङ्क मूल्यानि[], पूर्णाङ्क निम्नम्, पूर्णाङ्क उच्चम्) {
  यदि (निम्नम् < उच्चम्) { पूर्णाङ्क मध्यम् = विभाजय(मूल्यानि, निम्नम्, उच्चम्); शीघ्र_क्रमय(मूल्यानि, निम्नम्, मध्यम् - 1); शीघ्र_क्रमय(मूल्यानि, मध्यम् + 1, उच्चम्); }
}
पूर्णाङ्क मुख्यम्(शून्य) {
  पूर्णाङ्क मूल्यानि[] = {29, 3, 72, 14, 8, 55, 1, 40};
  शीघ्र_क्रमय(मूल्यानि, 0, 7);
  प्रति (पूर्णाङ्क अनुक्रमः = 0; अनुक्रमः < 8; अनुक्रमः++) प्रारूप_लिख("%d ", मूल्यानि[अनुक्रमः]);
  प्रारूप_लिख("\n");
  प्रत्यागम 0;
}
```
