## Detailed description

### Shaili guru: C in Sanskrit

Shaili guru is C written in the language: the language of systems, memory and the machine, for high school, undergraduate and GATE students. A program here goes through the script filter and Romenagri, the language's transducer gives C, and gcc compiles it; the names reach the binary in their Romenagri form.

**Program editing.** Edit the program in the next cell, then run it to save it.

```python
%%writefile program_guru.c
#अन्तर्भाव <stdio.h>
/* binary search in a sorted array */
पूर्णाङ्क द्विमान_अन्वेषणम्(पूर्णाङ्क मूल्यानि[], पूर्णाङ्क गणना_2, पूर्णाङ्क कुञ्जी) {
  पूर्णाङ्क निम्नम् = 0, उच्चम् = गणना_2 - 1;
  यावत् (निम्नम् <= उच्चम्) {
    पूर्णाङ्क मध्यम् = निम्नम् + (उच्चम् - निम्नम्) / 2;
    यदि (मूल्यानि[मध्यम्] == कुञ्जी) प्रत्यागम मध्यम्;
    यदि (मूल्यानि[मध्यम्] < कुञ्जी) निम्नम् = मध्यम् + 1; अन्यथा उच्चम् = मध्यम् - 1;
  }
  प्रत्यागम -1;
}
पूर्णाङ्क मुख्यम्(शून्य) {
  पूर्णाङ्क मूल्यानि[] = {2, 5, 8, 12, 16, 23, 38, 56, 72, 91};
  प्रारूप_लिख("%d %d\n", द्विमान_अन्वेषणम्(मूल्यानि, 10, 23), द्विमान_अन्वेषणम्(मूल्यानि, 10, 7));
  प्रत्यागम 0;
}
```

**Input editing.** Anything the program reads goes here.

```python
#@title Input editing
INPUT = """"""   # what the program reads, one value per line
open('input_guru.txt', 'w', encoding='utf-8').write(INPUT)
```

**The compilation flow.** The program as written, the script filter's Devanagari (for scripts that need one), the Romenagri form the compiler works on, and the host-language text.

```python
nb.show_stages(open('program_guru.c', encoding='utf-8').read(), 'sanskrit', 'guru')
```

**Program execution.**

```python
nb.build_and_run(open('program_guru.c', encoding='utf-8').read(), 'sanskrit', 'guru',
                 open('input_guru.txt', encoding='utf-8').read())
```

**From the host language (Std2Hin).** The English original of the same program, brought into Sanskrit by BURP.

```python
print(nb.from_host(open(REPO + '/samples/gate/algorithms/searching/binary_search.c', encoding='utf-8').read(), 'sanskrit', 'guru'))
```

### Shaili soochee: Python in Sanskrit

Shaili soochee is Python written in the language. It goes through the same path as C, with the language's Python transducer at the end, so a name has the same form in Python as in C.

**Program editing.** Edit the program in the next cell, then run it to save it.

```python
%%writefile program_soochee.py
# binary search, in Python
परिभाषा द्विमान_अन्वेषणम्(मूल्यानि, कुञ्जी):
    निम्नम्, उच्चम् = 0, दीर्घता(मूल्यानि) - 1
    यावत् निम्नम् <= उच्चम्:
        मध्यम् = (निम्नम् + उच्चम्) // 2
        यदि मूल्यानि[मध्यम्] == कुञ्जी:
            प्रत्यागम मध्यम्
        यदि मूल्यानि[मध्यम्] < कुञ्जी:
            निम्नम् = मध्यम् + 1
        अन्यथा:
            उच्चम् = मध्यम् - 1
    प्रत्यागम -1

मुद्रय(द्विमान_अन्वेषणम्([2, 5, 8, 12, 16, 23, 38], 23), द्विमान_अन्वेषणम्([2, 5, 8], 7))
```

**Input editing.** Anything the program reads goes here.

```python
#@title Input editing
INPUT = """"""   # what the program reads, one value per line
open('input_soochee.txt', 'w', encoding='utf-8').write(INPUT)
```

**The compilation flow.** The program as written, the script filter's Devanagari (for scripts that need one), the Romenagri form the compiler works on, and the host-language text.

```python
nb.show_stages(open('program_soochee.py', encoding='utf-8').read(), 'sanskrit', 'soochee')
```

**Program execution.**

```python
nb.build_and_run(open('program_soochee.py', encoding='utf-8').read(), 'sanskrit', 'soochee',
                 open('input_soochee.txt', encoding='utf-8').read())
```

**From the host language (Std2Hin).** The English original of the same program, brought into Sanskrit by BURP.

```python
print(nb.from_host(open(REPO + '/samples/gate/paradigms/binary_search.py', encoding='utf-8').read(), 'sanskrit', 'soochee'))
```
