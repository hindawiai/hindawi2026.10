```python
%%hindawi_c pashto
#ضمن <stdio.h>
/* quick sort */
فراغ بدلون(صحيح *لومړی, صحيح *دوهم) { صحيح لنډمهاله = *لومړی; *لومړی = *دوهم; *دوهم = لنډمهاله; }
صحيح برخه_کول(صحيح ارزښتونه[], صحيح ټیټ, صحيح لوړ) {
  صحيح محور = ارزښتونه[لوړ], شاخص = ټیټ - 1;
  لكل (صحيح لټون = ټیټ; لټون < لوړ; لټون++)
    إذا (ارزښتونه[لټون] <= محور) { شاخص++; بدلون(&ارزښتونه[شاخص], &ارزښتونه[لټون]); }
  بدلون(&ارزښتونه[شاخص + 1], &ارزښتونه[لوړ]);
  أرجع شاخص + 1;
}
فراغ چټک_ترتیبول(صحيح ارزښتونه[], صحيح ټیټ, صحيح لوړ) {
  إذا (ټیټ < لوړ) { صحيح منځ = برخه_کول(ارزښتونه, ټیټ, لوړ); چټک_ترتیبول(ارزښتونه, ټیټ, منځ - 1); چټک_ترتیبول(ارزښتونه, منځ + 1, لوړ); }
}
صحيح اصلي(فراغ) {
  صحيح ارزښتونه[] = {29, 3, 72, 14, 8, 55, 1, 40};
  چټک_ترتیبول(ارزښتونه, 0, 7);
  لكل (صحيح شاخص = 0; شاخص < 8; شاخص++) چاپ_کړه("%d ", ارزښتونه[شاخص]);
  چاپ_کړه("\n");
  أرجع 0;
}
```
