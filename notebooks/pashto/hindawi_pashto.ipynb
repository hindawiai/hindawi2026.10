## Detailed description

### Shaili guru: C in Pashto

Shaili guru is C written in the language: the language of systems, memory and the machine, for high school, undergraduate and GATE students. A program here goes through the script filter and Romenagri, the language's transducer gives C, and gcc compiles it; the names reach the binary in their Romenagri form.

**Program editing.** Edit the program in the next cell, then run it to save it.

```python
%%writefile program_guru.c
#ضمن <stdio.h>
/* binary search in a sorted array */
صحيح دوګونی_پلټنه(صحيح ارزښتونه[], صحيح شمېر, صحيح کیلي) {
  صحيح ټیټ = 0, لوړ = شمېر - 1;
  طالما (ټیټ <= لوړ) {
    صحيح منځ = ټیټ + (لوړ - ټیټ) / 2;
    إذا (ارزښتونه[منځ] == کیلي) أرجع منځ;
    إذا (ارزښتونه[منځ] < کیلي) ټیټ = منځ + 1; وإلا لوړ = منځ - 1;
  }
  أرجع -1;
}
صحيح اصلي(فراغ) {
  صحيح ارزښتونه[] = {2, 5, 8, 12, 16, 23, 38, 56, 72, 91};
  چاپ_کړه("%d %d\n", دوګونی_پلټنه(ارزښتونه, 10, 23), دوګونی_پلټنه(ارزښتونه, 10, 7));
  أرجع 0;
}
```

**Input editing.** Anything the program reads goes here.

```python
#@title Input editing
INPUT = """"""   # what the program reads, one value per line
open('input_guru.txt', 'w', encoding='utf-8').write(INPUT)
```

**The compilation flow.** The program as written, the script filter's Devanagari (for scripts that need one), the Romenagri form the compiler works on, and the host-language text.

```python
nb.show_stages(open('program_guru.c', encoding='utf-8').read(), 'pashto', 'guru')
```

**Program execution.**

```python
nb.build_and_run(open('program_guru.c', encoding='utf-8').read(), 'pashto', 'guru',
                 open('input_guru.txt', encoding='utf-8').read())
```

**From the host language (Std2Hin).** The English original of the same program, brought into Pashto by BURP.

```python
print(nb.from_host(open(REPO + '/samples/gate/algorithms/searching/binary_search.c', encoding='utf-8').read(), 'pashto', 'guru'))
```

### Shaili soochee: Python in Pashto

Shaili soochee is Python written in the language. It goes through the same path as C, with the language's Python transducer at the end, so a name has the same form in Python as in C.

**Program editing.** Edit the program in the next cell, then run it to save it.

```python
%%writefile program_soochee.py
# binary search, in Python
تعریف دوګونی_پلټنه(ارزښتونه, کیلي):
    ټیټ, لوړ = 0, اوږدوالی(ارزښتونه) - 1
    طالما ټیټ <= لوړ:
        منځ = (ټیټ + لوړ) // 2
        إذا ارزښتونه[منځ] == کیلي:
            أرجع منځ
        إذا ارزښتونه[منځ] < کیلي:
            ټیټ = منځ + 1
        وإلا:
            لوړ = منځ - 1
    أرجع -1

چاپ(دوګونی_پلټنه([2, 5, 8, 12, 16, 23, 38], 23), دوګونی_پلټنه([2, 5, 8], 7))
```

**Input editing.** Anything the program reads goes here.

```python
#@title Input editing
INPUT = """"""   # what the program reads, one value per line
open('input_soochee.txt', 'w', encoding='utf-8').write(INPUT)
```

**The compilation flow.** The program as written, the script filter's Devanagari (for scripts that need one), the Romenagri form the compiler works on, and the host-language text.

```python
nb.show_stages(open('program_soochee.py', encoding='utf-8').read(), 'pashto', 'soochee')
```

**Program execution.**

```python
nb.build_and_run(open('program_soochee.py', encoding='utf-8').read(), 'pashto', 'soochee',
                 open('input_soochee.txt', encoding='utf-8').read())
```

**From the host language (Std2Hin).** The English original of the same program, brought into Pashto by BURP.

```python
print(nb.from_host(open(REPO + '/samples/gate/paradigms/binary_search.py', encoding='utf-8').read(), 'pashto', 'soochee'))
```
