```python
%%hindawi_c punjabi_shahmukhi
#ضمن <stdio.h>
/* quick sort */
فراغ وٹاندرا(صحيح *پہلا, صحيح *دوجا) { صحيح عارضی = *پہلا; *پہلا = *دوجا; *دوجا = عارضی; }
صحيح حصے_کرو(صحيح مل[], صحيح نیواں, صحيح اچا) {
  صحيح دھرا = مل[اچا], اشاریہ = نیواں - 1;
  لكل (صحيح جانچ = نیواں; جانچ < اچا; جانچ++)
    إذا (مل[جانچ] <= دھرا) { اشاریہ++; وٹاندرا(&مل[اشاریہ], &مل[جانچ]); }
  وٹاندرا(&مل[اشاریہ + 1], &مل[اچا]);
  أرجع اشاریہ + 1;
}
فراغ تیز_ترتیب_دیو(صحيح مل[], صحيح نیواں, صحيح اچا) {
  إذا (نیواں < اچا) { صحيح وچکار = حصے_کرو(مل, نیواں, اچا); تیز_ترتیب_دیو(مل, نیواں, وچکار - 1); تیز_ترتیب_دیو(مل, وچکار + 1, اچا); }
}
صحيح مکھ(فراغ) {
  صحيح مل[] = {29, 3, 72, 14, 8, 55, 1, 40};
  تیز_ترتیب_دیو(مل, 0, 7);
  لكل (صحيح اشاریہ = 0; اشاریہ < 8; اشاریہ++) لکھو("%d ", مل[اشاریہ]);
  لکھو("\n");
  أرجع 0;
}
```
