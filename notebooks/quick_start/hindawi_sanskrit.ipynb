## Installation

Run this cell once. It fetches Hindawi, builds Chintamani's Romenagri tools and transducers, and loads the compilation path.

```python
import os, subprocess, sys, tempfile, pathlib
REPO = os.environ.get('HINDAWI_REPO') or os.path.abspath('hindawi2026.10')
if not os.path.isdir(REPO):
    subprocess.run(['git', 'clone', '-q', '--depth', '1', 'https://github.com/hindawiai/hindawi2026.10', REPO], check=True)
if 'google.colab' in sys.modules:
    subprocess.run('apt-get -qq install -y flex bison gawk > /dev/null', shell=True)
RMN = os.path.join(REPO, '.build/chintamani/Romenagri')
if not os.path.exists(os.path.join(RMN, 'rmn2acii')):
    subprocess.run(['bash', os.path.join(REPO, 'scripts/build.sh')], check=True)
os.environ['PATH'] = RMN + os.pathsep + os.environ['PATH']; os.environ['ROMENAGRI'] = RMN
sys.path.insert(0, os.path.join(REPO, 'lang/hindawi/tools'))
import hindawi_chain as hc
LANG = 'sanskrit'
def stages(src, fwd='h2c'):
    for n, t in hc.stages(src, LANG, fwd):
        print('===== ' + n); print(t)
def build(src):
    d = pathlib.Path(tempfile.mkdtemp(prefix='hindawi-')); f = d / 'program.c'; f.write_text(src, encoding='utf-8')
    env = dict(os.environ, HINDAWI_LANG=LANG, HINDAWI_NATIVE_SOURCE='1')
    r = subprocess.run([os.path.join(REPO, 'lang/hindawi/hindawicc'), '-O2', '-o', str(d / 'program'), str(f), '-lm'],
                       capture_output=True, text=True, env=env)
    return (d / 'program' if r.returncode == 0 else None), r.stderr
def run(exe, stdin=''):
    return subprocess.run([str(exe)], input=stdin, capture_output=True, text=True, timeout=20).stdout
def run_py(src):
    return subprocess.run([sys.executable, '-c', hc.to_py(src, LANG)], capture_output=True, text=True, timeout=20)
def to_language(c_text):
    d = pathlib.Path(tempfile.mkdtemp(prefix='hindawi-')); (d / 'src').mkdir(); (d / 'src/program.c').write_text(c_text, encoding='utf-8')
    subprocess.run([sys.executable, os.path.join(REPO, 'lang/hindawi/burp/burp.py'), 'ingest', str(d / 'src'),
                    '--langs', LANG, '--out', str(d / 'n')], check=True, capture_output=True)
    return (d / 'n' / LANG / 'program.c').read_text(encoding='utf-8')
print('ready:', REPO)
```

## Practice area

Write in Sanskrit. **Build** compiles through the original pipeline; **Run** runs it with the input on the left; **Stages** shows every step; **To C** shows the C the compiler gets; **From C** turns C in the editor into the language; **New** clears everything.

```python
import ipywidgets as w
from IPython.display import display
program = w.Textarea(value='#अन्तर्भाव <stdio.h>\n/* binary search in a sorted array */\nपूर्णाङ्क द्विमान_अन्वेषणम्(पूर्णाङ्क मूल्यानि[], पूर्णाङ्क गणना_2, पूर्णाङ्क कुञ्जी) {\n  पूर्णाङ्क निम्नम् = 0, उच्चम् = गणना_2 - 1;\n  यावत् (निम्नम् <= उच्चम्) {\n    पूर्णाङ्क मध्यम् = निम्नम् + (उच्चम् - निम्नम्) / 2;\n    यदि (मूल्यानि[मध्यम्] == कुञ्जी) प्रत्यागम मध्यम्;\n    यदि (मूल्यानि[मध्यम्] < कुञ्जी) निम्नम् = मध्यम् + 1; अन्यथा उच्चम् = मध्यम् - 1;\n  }\n  प्रत्यागम -1;\n}\nपूर्णाङ्क मुख्यम्(शून्य) {\n  पूर्णाङ्क मूल्यानि[] = {2, 5, 8, 12, 16, 23, 38, 56, 72, 91};\n  प्रारूप_लिख("%d %d\\n", द्विमान_अन्वेषणम्(मूल्यानि, 10, 23), द्विमान_अन्वेषणम्(मूल्यानि, 10, 7));\n  प्रत्यागम 0;\n}\n', layout=w.Layout(width='98%', height='300px'))
inp = w.Textarea(layout=w.Layout(width='48%', height='110px'), placeholder='input')
out = w.Textarea(layout=w.Layout(width='48%', height='110px'), placeholder='output')
view = w.Output(layout=w.Layout(width='98%', border='1px solid #ccc'))
state = {'exe': None}
def do_build(_):
    state['exe'], err = build(program.value); out.value = 'built' if state['exe'] else err
def do_run(_):
    out.value = run(state['exe'], inp.value) if state['exe'] else 'Build first'
def do_stages(_):
    view.clear_output()
    with view: stages(program.value)
def do_to_c(_):
    out.value = hc.to_c(program.value, LANG)
def do_from_c(_):
    program.value = to_language(program.value)
def do_new(_):
    program.value = inp.value = out.value = ''; state['exe'] = None; view.clear_output()
acts = [('Build', do_build), ('Run', do_run), ('Stages', do_stages), ('To C', do_to_c), ('From C', do_from_c), ('New', do_new)]
bar = w.HBox([w.Button(description=d) for d, _ in acts])
for b, (_, f) in zip(bar.children, acts): b.on_click(f)
display(w.VBox([bar, program, w.HBox([inp, out]), view]))
```

## Shaili guru (C): every stage

A binary search written in Sanskrit. Each stage below is what the compiler really gets: the script filter's Devanagari (scripts other than Devanagari), the Romenagri form it compiles, and the C its transducer gives.

```python
SRC_C = '#अन्तर्भाव <stdio.h>\n/* binary search in a sorted array */\nपूर्णाङ्क द्विमान_अन्वेषणम्(पूर्णाङ्क मूल्यानि[], पूर्णाङ्क गणना_2, पूर्णाङ्क कुञ्जी) {\n  पूर्णाङ्क निम्नम् = 0, उच्चम् = गणना_2 - 1;\n  यावत् (निम्नम् <= उच्चम्) {\n    पूर्णाङ्क मध्यम् = निम्नम् + (उच्चम् - निम्नम्) / 2;\n    यदि (मूल्यानि[मध्यम्] == कुञ्जी) प्रत्यागम मध्यम्;\n    यदि (मूल्यानि[मध्यम्] < कुञ्जी) निम्नम् = मध्यम् + 1; अन्यथा उच्चम् = मध्यम् - 1;\n  }\n  प्रत्यागम -1;\n}\nपूर्णाङ्क मुख्यम्(शून्य) {\n  पूर्णाङ्क मूल्यानि[] = {2, 5, 8, 12, 16, 23, 38, 56, 72, 91};\n  प्रारूप_लिख("%d %d\\n", द्विमान_अन्वेषणम्(मूल्यानि, 10, 23), द्विमान_अन्वेषणम्(मूल्यानि, 10, 7));\n  प्रत्यागम 0;\n}\n'
stages(SRC_C)
```

## Compile, run, and compare with the English program

```python
ENGLISH_C = '#include <stdio.h>\n/* binary search in a sorted array */\nint binary_search(int values[], int count, int key) {\n  int low = 0, high = count - 1;\n  while (low <= high) {\n    int middle = low + (high - low) / 2;\n    if (values[middle] == key) return middle;\n    if (values[middle] < key) low = middle + 1; else high = middle - 1;\n  }\n  return -1;\n}\nint main(void) {\n  int values[] = {2, 5, 8, 12, 16, 23, 38, 56, 72, 91};\n  printf("%d %d\\n", binary_search(values, 10, 23), binary_search(values, 10, 7));\n  return 0;\n}\n'
exe, err = build(SRC_C)
d = pathlib.Path(tempfile.mkdtemp()); (d / 'e.c').write_text(ENGLISH_C)
subprocess.run(['gcc', '-O2', '-o', str(d / 'e'), str(d / 'e.c')], check=True)
mine, english = (run(exe) if exe else err), run(d / 'e')
print(mine); print('same as English:', mine == english)
```

## Shaili soochee (Python): every stage, then run

```python
SRC_PY = '# binary search, in Python\nपरिभाषा द्विमान_अन्वेषणम्(मूल्यानि, कुञ्जी):\n    निम्नम्, उच्चम् = 0, दीर्घता(मूल्यानि) - 1\n    यावत् निम्नम् <= उच्चम्:\n        मध्यम् = (निम्नम् + उच्चम्) // 2\n        यदि मूल्यानि[मध्यम्] == कुञ्जी:\n            प्रत्यागम मध्यम्\n        यदि मूल्यानि[मध्यम्] < कुञ्जी:\n            निम्नम् = मध्यम् + 1\n        अन्यथा:\n            उच्चम् = मध्यम् - 1\n    प्रत्यागम -1\n\nमुद्रय(द्विमान_अन्वेषणम्([2, 5, 8, 12, 16, 23, 38], 23), द्विमान_अन्वेषणम्([2, 5, 8], 7))\n'
stages(SRC_PY, 'h2py')
r = run_py(SRC_PY)
print(r.stdout or r.stderr)
```

## From English to Sanskrit (BURP)

```python
print(to_language(ENGLISH_C))
```
