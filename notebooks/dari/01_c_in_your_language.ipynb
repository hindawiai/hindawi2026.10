```python
%%hindawi_c dari
#ضمن <stdio.h>
/* quick sort */
فراغ جابجا(صحيح *اولی, صحيح *دومی) { صحيح موقت = *اولی; *اولی = *دومی; *دومی = موقت; }
صحيح افراز(صحيح مقادیر[], صحيح کم, صحيح بالا) {
  صحيح محور = مقادیر[بالا], شاخص = کم - 1;
  لكل (صحيح پیمایش = کم; پیمایش < بالا; پیمایش++)
    إذا (مقادیر[پیمایش] <= محور) { شاخص++; جابجا(&مقادیر[شاخص], &مقادیر[پیمایش]); }
  جابجا(&مقادیر[شاخص + 1], &مقادیر[بالا]);
  أرجع شاخص + 1;
}
فراغ سریع_مرتب_سازی(صحيح مقادیر[], صحيح کم, صحيح بالا) {
  إذا (کم < بالا) { صحيح میانه = افراز(مقادیر, کم, بالا); سریع_مرتب_سازی(مقادیر, کم, میانه - 1); سریع_مرتب_سازی(مقادیر, میانه + 1, بالا); }
}
صحيح اصلی(فراغ) {
  صحيح مقادیر[] = {29, 3, 72, 14, 8, 55, 1, 40};
  سریع_مرتب_سازی(مقادیر, 0, 7);
  لكل (صحيح شاخص = 0; شاخص < 8; شاخص++) چاپ_کن("%d ", مقادیر[شاخص]);
  چاپ_کن("\n");
  أرجع 0;
}
```
