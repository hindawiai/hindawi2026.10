```python
%%hindawi_py english
def binary_search(values, key):
    low, high = 0, len(values) - 1
    while low <= high:
        middle = (low + high) // 2
        if values[middle] == key:
            return middle
        if values[middle] < key:
            low = middle + 1
        else:
            high = middle - 1
    return None

for key in [23, 7]:
    print(key, binary_search([2, 5, 8, 12, 16, 23, 38], key))
```
